## Setup

````python
import pandas as pd
import statsmodels.formula.api as smf

# Our class data from Week 1, with the finger-length typos we fixed in Week 3
# already corrected (finger_length_corrected).
data_path = "https://raw.githubusercontent.com/stevenolanecon/7002LBSAI/main/data/week3_reaction_times.csv"
df = pd.read_csv(data_path)
df['finger_mm'] = df['finger_length_corrected']

# Every regression in this notebook is fit on one row per student (their
# mean reaction time), not on the raw trial-level rows – same convention
# as every finger-length regression in this week's slides.
student_summary = (
    df.groupby('name')
    .agg(mean_rt=('reaction_time_ms', 'mean'), finger_mm=('finger_mm', 'first'))
    .reset_index()
)
student_summary
````
