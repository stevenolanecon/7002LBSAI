## Setup

````python
import pandas as pd
import statsmodels.formula.api as smf

# Live in-class data – the same Google Sheet the Week 1 tool writes to.
data_path = "https://docs.google.com/spreadsheets/d/e/2PACX-1vRWIOZ03GpUJgqvGVmI9cih169mMq0ajLMj6a0VTd2tytB3K5BmX5AUiKQ8FKIL6KUKEcgb-1LR-1eM/pub?gid=0&single=true&output=csv"
df = pd.read_csv(data_path)

if 'name' in df.columns and 'student_name' not in df.columns:
    df = df.rename(columns={'name': 'student_name'})
df['finger_mm'] = pd.to_numeric(df['finger_mm'], errors='coerce')

# Every regression in this notebook is fit on one row per student (their
# mean reaction time), not on the raw trial-level rows – same convention
# as every finger-length regression in this week's slides.
student_summary = (
    df.groupby('student_name')
    .agg(mean_rt=('reaction_time_ms', 'mean'), finger_mm=('finger_mm', 'first'))
    .reset_index()
    .dropna(subset=['finger_mm'])
)
student_summary
````
