> **Syntax hint – bin-scatter.** You met bin-scatter conceptually in Week 3. To build one in Python, split `x` into equal-sized groups with `pd.qcut()`, then take the mean of `y` within each group:
>
> ```python
> df['x_bin'] = pd.qcut(df['x'], q=NUMBER_OF_BINS)
> bin_means = df.groupby('x_bin')['y'].mean()
> bin_means.plot(kind='bar')
> ```
> `pd.qcut` splits into bins with roughly equal numbers of observations; `pd.cut` splits into bins of equal width instead. Either is a reasonable choice here – think about which suits this variable better.

> **Syntax hint – LOWESS.** `seaborn` can fit and plot a LOWESS curve directly:
>
> ```python
> sns.regplot(x='x', y='y', data=df, lowess=True, scatter=True)
> ```
>
> If you want the fitted values themselves rather than just a plot, `statsmodels` has its own `lowess` function:
>
> ```python
> from statsmodels.nonparametric.smoothers_lowess import lowess
> smoothed = lowess(df['y'], df['x'])  # returns an array of (x, fitted y) pairs, sorted by x
> ```

> **Syntax hint – residuals and fitted values.** Once you've fit a model with `smf.ols(...).fit()`, the fitted model object stores both the fitted values and the residuals as attributes:
>
> ```python
> model = smf.ols('y ~ x', data=df).fit()
> df['fitted'] = model.fittedvalues
> df['resid'] = model.resid
> ```
>
> A large **negative** residual means the actual price is much lower than the model predicts for a car of that age – i.e. a candidate good deal. Sort by `resid` to find them.
